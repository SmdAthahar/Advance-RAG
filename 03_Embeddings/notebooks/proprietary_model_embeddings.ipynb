```python
from dotenv import load_dotenv
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.document_loaders import PyPDFLoader
from langchain_openai import OpenAIEmbeddings
```

```python
load_dotenv()
```

```python
import os

EMB_MODEL_LARGE = OpenAIEmbeddings(
    model="text-embedding-3-large",
    api_key=os.environ["EMB_API_KEY"],
    base_url="https://api.euron.one/api/v1/euri",
)

EMB_MODEL_SMALL = OpenAIEmbeddings(
    model="text-embedding-3-small",
    api_key=os.environ["EMB_API_KEY"],
    base_url="https://api.euron.one/api/v1/euri",
)
```

```python
# create embeddings from user query

query = "What is Openclaw/Moltbot and what are the major security concerns regarding this tool"

embeddings_large = EMB_MODEL_LARGE.embed_query(
    text=query
)
```

```python
embeddings_small = EMB_MODEL_SMALL.embed_query(
    text=query
)

len(embeddings_small)
```

```python
loader = PyPDFLoader(file_path="../Openclaw_Research_Report.pdf")

docs = loader.load()

len(docs)
```

```python
chunker = RecursiveCharacterTextSplitter(
    chunk_size=300,
    chunk_overlap=50
)

chunks = chunker.split_documents(docs)

len(chunks)
```

```python
# embed documents

text_documents = [doc.page_content for doc in chunks]

print(len(text_documents))

document_embeddings = EMB_MODEL_LARGE.embed_documents(
    texts=text_documents
)
```

```python
len(document_embeddings)
```

```python
document_embeddings[0:3]
```

```python
# change the dimensions

embedder_large_custom_dim = OpenAIEmbeddings(
    model="text-embedding-3-large",
    dimensions=256,
    api_key=os.environ["EMB_API_KEY"],
    base_url="https://api.euron.one/api/v1/euri"
)
```

```python
query_embeddings = embedder_large_custom_dim.embed_query(text=query)
```
