```python
from langchain_experimental.text_splitter import SemanticChunker
from langchain_openai import OpenAIEmbeddings
from dotenv import load_dotenv
import os
```

```python
text = """Artificial intelligence is transforming technology and shaping the future.
Machine learning algorithms are becoming more sophisticated every day.
Deep learning models can now process vast amounts of data efficiently.
Neural networks are inspired by the human brain's structure.
The best pasta recipes include fresh ingredients and proper cooking techniques.
Italian cuisine emphasizes quality olive oil and regional cheeses.
Authentic carbonara uses guanciale, eggs, pecorino romano, and black pepper.
Cooking pasta al dente ensures the best texture and flavor.
Climate change is affecting ecosystems worldwide.
Rising temperatures are causing glaciers to melt at unprecedented rates.
Scientists warn that immediate action is needed to reduce carbon emissions.
Renewable energy sources offer hope for a sustainable future."""
```

```python
load_dotenv()
```

```python
embeddings = OpenAIEmbeddings(
    model="text-embedding-3-small",
    api_key=os.environ["EURI_API_KEY"],
    base_url="https://api.euron.one/api/v1/euri",
)
```

```python
# create semantic chunker

chunker = SemanticChunker(
    embeddings = embeddings,
    breakpoint_threshold_type = "standard_deviation",
    breakpoint_threshold_amount = 0.8
)
```

```python
# split the text

chunks = chunker.split_text(text)

print(chunks)
```

```python
from termcolor import COLORS, colored
from random import choice
```

```python
def display_chunks(chunks):
    colors_list = list(COLORS.keys())[2:8]
    print(f"Total Number of Chunks: {len(chunks)}")
    
    for num, chunk in enumerate(chunks, 1):
        print(f"Chunk {num}: Length {len(chunk)} chars")
        print(colored(text=chunk, color=choice(colors_list)), end="\n\n")
```

```python
display_chunks(chunks)
```

```python
text8 = """
Artificial intelligence is changing the technology industry.

Machine learning algorithms are becoming increasingly powerful.

Deep learning models can process large amounts of information.

Neural networks are used in computer vision and language processing.

Pizza is one of the most popular foods in the world.

Italian pizza traditionally uses simple ingredients.

Fresh tomatoes, mozzarella, and basil are common ingredients.

Climate change is causing major environmental changes.

Glaciers are melting because global temperatures are rising.

Renewable energy can help reduce greenhouse gas emissions.
"""
```

```python
chunker_a = SemanticChunker(
    embeddings=embeddings,
    breakpoint_threshold_type="percentile",
    breakpoint_threshold_amount=80
)
```

```python
chunker_b = SemanticChunker(
    embeddings=embeddings,
    breakpoint_threshold_type="standard_deviation",
    breakpoint_threshold_amount=0.8
)
```

```python
chunker_c = SemanticChunker(
    embeddings=embeddings,
    breakpoint_threshold_type="interquartile",
    breakpoint_threshold_amount=1.5
)
```

```python
chunker_d = SemanticChunker(
    embeddings=embeddings,
    breakpoint_threshold_type="gradient",
    breakpoint_threshold_amount=0.5
)
```

```python
for name, chunker in [
    ("Percentile", chunker_a),
    ("Standard Deviation", chunker_b),
    ("Interquartile", chunker_c),
    ("Gradient", chunker_d)
]:
    
    chunks = chunker.split_text(text8)
    
    print(f"\n\n{'='*20} {name} {'='*20}")
    print(f"Number of chunks: {len(chunks)}")
    
    for i, chunk in enumerate(chunks):
        print(f"\n--- Chunk {i+1} ---")
        print(chunk)
```
